1. Dada uma lista de números, encontre o menor valor.

```python
lista = []

for i in range(5):
    numero = int(input(f"Digite {i+1}º número: "))
    lista.append(numero)

menor = lista[0]
for _ in lista:
    if _ < menor:
        menor = _

print(menor)
```

2. Crie uma função que receba uma lista e retorne a soma dos elementos pares.

```python
lista = []

for i in range(5):
    numero = int(input())
    lista.append(numero)

soma = 0
for _ in lista:
    if _ % 2 == 0:
        soma += _

print(soma)
```

3. Crie uma lista contendo os quadrados dos números de outra lista.

```python
lista = []

for i in range(5):
    numero = int(input("Número: "))
    lista.append(numero)

quadrado = []

for _ in lista:
    resultado = _ **2
    quadrado.append(resultado)

print(lista)
print(quadrado)
```

4. Encontre a frequência de cada elemento em uma lista.

```python
lista = []
valores = []
contagem = []

for i in range(6):
    numero = int(input("Número: "))
    lista.append(numero)

for n in lista:

    if n in valores:
        i = valores.index(n)
        contagem[i] += 1
    
    else:
        valores.append(n)
        contagem.append(1)

print(valores)
print(contagem)
```

5. Remova os duplicados de uma lista, mantendo a ordem original.

```python
lista = []

for i in range(5):

    numero = int(input('Número: '))
    lista.append(numero)

resultado = []

for n in lista:
    if n not in resultado:
        resultado.append(n)

print(lista)
print(resultado)
```

6. Inverta uma lista sem usar a função embutida reverse().

```python
lista = []

for i in range(5):
    numero = int(input("Número: "))
    lista.append(numero)

inverso = lista[::-1]

print(inverso)
```

7. Conte quantos elementos em uma lista são maiores que um valor dado.

```python
valor = 5
lista = []
maior = 0

for i in range(5):
    numero = int(input("Número: "))
    lista.append(numero)

for _ in lista:
    if _ > valor:
        maior += 1

print(maior)
```

8. Faça uma lista com os elementos da primeira lista ao quadrado, mas somente se forem ímpares.

```python
lista = []
imparesQuadrados = []

for i in range(5):
    numero = int(input("Número: "))
    lista.append(numero)

for _ in lista:
    if _ % 2 != 0:
        quadrado = _ ** 2
        imparesQuadrados.append(quadrado)

print(imparesQuadrados)
```

9. Encontre o índice do primeiro valor que é maior que um determinado número.

```python
lista = []
limite = 6
indice = -1

for i in range(5):
    numero = int(input("Número: "))
    lista.append(numero)

for _ in range(len(lista)):
    if lista[_] > limite:
        indice = _
        break

print(lista)
print(indice)
```

10. Dada uma lista, intercale os valores de duas listas (ou seja, um de uma, um da outra, alternadamente).

```python
lista1 = [1,2,3]
lista2 = [4,5,6]

resultado = []

for i in range(len(lista1)):
    resultado.append(lista1[i])
    resultado.append(lista2[i])

print(resultado)
```
